```python
unemployment_data = pd.read_csv(os.path.join(DATA_DIR, 'nigeria_unemployment_missing_data.csv'))
unemployment_data.head()
```

```python
# To check for missing values
import pandas as pd
unemployment_data.isnull().sum()
```

```python
unemployment_data.shape
```

```python
# to preserve the original data while we edited another copy
unemployment_data_1 = unemployment_data.copy()
unemployment_data_1.head()
```

```python
# To take away all the missing values
unemployment_data_1.dropna(inplace=True)
# to get the new size of the data
unemployment_data_1.shape

```

```python
# to fix the years of experience columns with the average years of experience
# calculate the average years of experience
average_experience = unemployment_data['Years_Of_Experience'].mean()
average_experience
```

```python
## categorical patching( mode vs. unknown)
#strategy for dealing with missing education level data: flag as unknown or unspecified
# to check for categorical data under Education level
unemployment_data['Education_Level'].unique()
```

```python
# to replace the 'nan' with unspecified
unemployment_data['Education_Level'] = unemployment_data['Education_Level'].fillna('Unspecified')
unemployment_data['Education_Level'].unique()
```

```python
# The group based imputation
#calculate the average of the median within a specific group
#Then fill in accordingly
#Calculate the average experience per employment status group
group_experience = unemployment_data.groupby('Employment_Status')['Years_Of_Experience'].mean().round(2)
group_experience
```

```python
#Then fill in the average accordingly
unemployment_data['Years_Of_Experience'] = unemployment_data['Years_Of_Experience'].fillna(group_experience). inplace = True
```
