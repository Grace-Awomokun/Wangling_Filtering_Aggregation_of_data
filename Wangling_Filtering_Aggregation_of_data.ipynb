```python
# Data wangling is the process of taking messy data and transforming it into an organize format
#The superpower of data wangling is data filtering which is very important in
# ...machine learning workflow
# scope and focus can be easily distracted
#filtering allow us to remove logical impossibilities
# targetted slicing, by extracting out only the dataset needed
```

```python
# dataset : Nigerian nursing mothers
# Scenario 1:
#we want to know the health challenges faced by the poorest mothers living in rural area

# package of interest
import pandas as pd
```

```python
# upload the dataset of interest
health_data = pd.read_csv('/nigeria_nursing_mothers_healthcare.csv')
health_data.head()
```

```python
# create a new dataset from the existing one ( called subsetting )
targeted_mothers = health_data[(health_data['Residence'] == 'Rural') & (health_data['Wealth_Quintile'] == 'Poorest')]
```

```python
print(f"original dataset size:", len (health_data))
print(f"New dataset size:", len(targeted_mothers))
```
